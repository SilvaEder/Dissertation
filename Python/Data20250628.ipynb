```python
##IMPORTING LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Backup/202506211647.csv', header=None).T
df.columns = ["n", "x1", "x2", "x3", "x4","u","Latency", "t", "k"]
df.head()
```

```python
# Sampling period
T = 0.05  # seconds

# Number of samples
n = len(df)

# Generate time vector
t = np.arange(n) * T

# Create figure with GridSpec
fig = plt.figure(figsize=(10, 8))

gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 1])

axs = [
    fig.add_subplot(gs[0, 0]),
    fig.add_subplot(gs[0, 1]),
    fig.add_subplot(gs[1, 0]),
    fig.add_subplot(gs[1, 1]),
    fig.add_subplot(gs[2, :])
]

# State configurations
states = ["x1", "x2", "x3", "x4"]

labels = [
    "Position (m)",
    "Velocity (m/s)",
    "Angle (rad)",
    "Angular velocity (rad/s)"
]

limits = [0.30, 0.30, 0.20, 0.20]

ylims = [
    (-0.35, 0.35),
    (-0.35, 0.35),
    (-0.30, 0.30),
    (-0.30, 0.30)
]

# Plot states
for i, state in enumerate(states):

    ax = axs[i]
    limit = limits[i]

    # State signal
    ax.plot(
        t,
        df[state],
        label=fr"$x_{i+1}$",
        color="#0072B2"
    )

    # Limits
    ax.axhline(
        -limit,
        color="red",
        linestyle=":",
        linewidth=1.2,
        label="Limits"
    )

    ax.axhline(
        limit,
        color="red",
        linestyle=":",
        linewidth=1.2
    )

    # Reference
    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1,
        label="Reference"
    )

    # Valid operating region
    ax.axhspan(
        -limit,
        limit,
        facecolor="#0072B2",
        alpha=0.12,
        zorder=0
    )

    ax.set_title(fr"State $x_{i+1}$")
    ax.set_ylabel(labels[i])
    ax.set_ylim(ylims[i])
    ax.legend()


# Control input u
ax = axs[4]

ax.plot(
    t,
    df["u"],
    label=r"$u$",
    color="#F0E442"
)

# Control limits
ax.axhline(
    -0.30,
    color="red",
    linestyle=":",
    linewidth=1.2,
    label="Limits"
)

ax.axhline(
    0.30,
    color="red",
    linestyle=":",
    linewidth=1.2
)

# Reference
ax.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
    label="Reference"
)

# Valid operating region
ax.axhspan(
    -0.30,
    0.30,
    facecolor="#0072B2",
    alpha=0.12,
    zorder=0
)

ax.set_title(r"Control Input $u$")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Control signal")
ax.set_ylim([-0.40, 0.40])
ax.legend()

# Remove top and right borders
for ax in axs:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# Time labels
axs[2].set_xlabel("Time (s)")
axs[3].set_xlabel("Time (s)")

# Adjust layout
fig.tight_layout()

plt.show()
```

```python
# Sampling period
T = 0.05  # seconds

# Number of samples
n = len(df)

# Generate time vector
t = np.arange(n) * T

# Create figure with GridSpec
fig = plt.figure(figsize=(10, 8))

gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 1])

axs = [
    fig.add_subplot(gs[0, 0]),
    fig.add_subplot(gs[0, 1]),
    fig.add_subplot(gs[1, 0]),
    fig.add_subplot(gs[1, 1]),
    fig.add_subplot(gs[2, :])
]

# State configurations
states = ["x1", "x2", "x3", "x4"]

labels = [
    "Position (m)",
    "Velocity (m/s)",
    "Angle (rad)",
    "Angular velocity (rad/s)"
]

limits = [0.30, 0.30, 0.20, 0.20]

ylims = [
    (-0.35, 0.35),
    (-0.35, 0.35),
    (-0.30, 0.30),
    (-0.30, 0.30)
]

# Plot states
for i, state in enumerate(states):

    ax = axs[i]
    limit = limits[i]

    # State signal
    ax.plot(
        t,
        df[state],
        label=fr"$x_{i+1}$",
        color="#0072B2"
    )

    # Limits
    ax.axhline(
        -limit,
        color="red",
        linestyle=":",
        linewidth=1.2,
        label="Limits"
    )

    ax.axhline(
        limit,
        color="red",
        linestyle=":",
        linewidth=1.2
    )

    # Reference
    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1,
        label="Reference"
    )

    # Valid operating region
    ax.axhspan(
        -limit,
        limit,
        facecolor="#0072B2",
        alpha=0.12,
        zorder=0
    )

    ax.set_title(fr"State $x_{i+1}$")
    ax.set_ylabel(labels[i])
    ax.set_ylim(ylims[i])
    ax.legend()


# Control input u
ax = axs[4]

ax.plot(
    t,
    df["u"],
    label=r"$u$",
    color="#F0E442"
)

# Control limits
ax.axhline(
    -0.30,
    color="red",
    linestyle=":",
    linewidth=1.2,
    label="Limits"
)

ax.axhline(
    0.30,
    color="red",
    linestyle=":",
    linewidth=1.2
)

# Reference
ax.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
    label="Reference"
)

# Valid operating region
ax.axhspan(
    -0.30,
    0.30,
    facecolor="#0072B2",
    alpha=0.12,
    zorder=0
)

ax.set_title(r"Control Input $u$")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Control signal")
ax.set_ylim([-0.40, 0.40])
ax.legend()

# Remove top and right borders
for ax in axs:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# Time labels
axs[2].set_xlabel("Time (s)")
axs[3].set_xlabel("Time (s)")

# Adjust layout
fig.tight_layout()

plt.show()
```
