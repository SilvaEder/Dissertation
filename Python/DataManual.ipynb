```python
##IMPORTING LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Backup/202507121850-Manual.csv', header=None).T
df.columns = ["n", "x1", "x2", "x3", "x4","u","Latency", "t", "k"]
df.head()
```

```python
# Sampling period
T = 0.05  # seconds

# Number of samples
n = len(df)

# Generate time vector
t = np.arange(n) * T

# Create figure with GridSpec
fig = plt.figure(figsize=(10, 8))

gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 1])

axs = [
    fig.add_subplot(gs[0, 0]),
    fig.add_subplot(gs[0, 1]),
    fig.add_subplot(gs[1, 0]),
    fig.add_subplot(gs[1, 1]),
    fig.add_subplot(gs[2, :])
]

# States x1, x2, x3, x4
states = ["x1", "x2", "x3", "x4"]

for i, state in enumerate(states):

    axs[i].plot(
        t,
        df[state],
        label=fr"${state[0]}_{state[1]}$",
        color="#0072B2"
    )

    axs[i].set_title(fr"State $x_{i+1}$")
    axs[i].legend()

# Control input u
axs[4].plot(
    t,
    df["u"],
    label=r"$u$",
    color="#F0E442"
)

axs[4].set_title(r"Control Input $u$")
axs[4].set_xlabel("Time (s)")
axs[4].legend()

# Remove top and right borders
for ax in axs:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# Adjust layout
fig.tight_layout()

plt.show()
```
