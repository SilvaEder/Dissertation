```python
def MovingAverage(position):
    """
    CALCULATES ACCELERATION FROM POSITION AND APPLIES A MOVING AVERAGE.
    
    ARGS:
    POSITION: CURRENT POSITION.
    
    RETURNS:
    AVERAGE ACCELERATION.
    """
    x_previous.append(position)
    x_previous.pop(0)

    # CALCULATE THE INTENSITY OF THE INTENSITY
    acel = ((x_previous[2] - 2 * x_previous[1] + x_previous[0])) /t**2

    # ADD ACCELERATION TO HISTORY LIST
    acel_previous.append(acel)
    acel_previous.pop(0)

    # CALCULATES THE MOVING AVERAGE OF ACCELERATION
    avr_acel = sum(acel_previous) / len(acel_previous)
    return avr_acel    
```
