```python
def Phi(x):
    Phi_function = np.array([
        [x[0][0]], 
        [x[1][0]], 
        [x[2][0]], 
        [x[3][0]], 
        [x[0][0]**2], 
        [x[1][0]**2], 
        [x[2][0]**2], 
        [x[3][0]**2], 
        [x[0][0] * x[1][0]], 
        [x[0][0] * x[2][0]], 
        [x[0][0] * x[3][0]], 
        [x[1][0] * x[2][0]], 
        [x[1][0] * x[3][0]], 
        [x[2][0] * x[3][0]], 
        [x[0][0]**3], 
        [x[1][0]**3], 
        [x[2][0]**3], 
        [x[3][0]**3], 
        [x[0][0]**2 * x[1][0]], 
        [x[0][0]**2 * x[2][0]], 
        [x[0][0]**2 * x[3][0]], 
        [x[1][0]**2 * x[0][0]], 
        [x[1][0]**2 * x[2][0]], 
        [x[1][0]**2 * x[3][0]], 
        [x[2][0]**2 * x[0][0]], 
        [x[2][0]**2 * x[1][0]], 
        [x[2][0]**2 * x[3][0]], 
        [x[3][0]**2 * x[0][0]], 
        [x[3][0]**2 * x[1][0]], 
        [x[3][0]**2 * x[2][0]]])
    return Phi_function
```
